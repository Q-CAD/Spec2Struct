```python
import ase.build
import json
import ase.io
from tqdm import tqdm
from glob import glob
from pathlib import Path
```

```python
def save_to_json(cifs_path, output_path, data_type="train", supercell=False, n_cell=3):
    cifs = glob(str(cifs_path / "*.cif"))
    assert len(cifs) > 0

    atoms_dict = {
        Path(cif).stem: ase.io.read(cif)
        for cif in tqdm(cifs, desc=f"Reading {type} CIFs")
    }

    atoms_list = []

    for name, atoms in tqdm(atoms_dict.items(), desc=f"Converting {type} CIFs to JSON"):
        if supercell:
            # Convert to supercell if needed
            superatoms = ase.build.make_supercell(
                atoms, 
                [[n_cell, 0, 0], [0, n_cell, 0], [0, 0, n_cell]]
            )

        atoms_dict = {
            'structure_id': name,
            'positions': superatoms.get_positions().tolist(),
            'cell': superatoms.get_cell().tolist(),
            'atomic_numbers': superatoms.get_atomic_numbers().tolist(),
            'y': [0.0]
        }

        atoms_list.append(atoms_dict)
    
    save_fname = output_path / f"{data_type}.json"
    with open(save_fname, "w") as f:
        json.dump(atoms_list, f)
```

```python
path = "/global/cfs/projectdirs/m3641/Shared/Materials_datasets/CDVAE/perov_5/raw_train/"
output_path = "data/perov5_supercell_json/"

save_to_json(Path(path), Path(output_path), data_type="train", supercell=True, n_cell=2)
```

```python
path = "/global/cfs/projectdirs/m3641/Shared/Materials_datasets/CDVAE/perov_5/raw_val/"
output_path = "data/perov5_supercell_json/"

save_to_json(Path(path), Path(output_path), data_type="val", supercell=True, n_cell=2)
```
